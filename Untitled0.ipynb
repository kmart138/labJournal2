```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 1, 100)

def compute_exponential(x):
    return np.exp(x)

y = compute_exponential(x)

plt.figure()
plt.plot(x, y)
plt.xlabel("Time [milliseconds]")
plt.ylabel("Awesomeness")
plt.title("Exponential Growth of Awesomeness")
plt.grid(True)
plt.savefig("exponential_plot.pdf")
plt.show()

from google.colab import files
files.download("exponential_plot.pdf")

```
